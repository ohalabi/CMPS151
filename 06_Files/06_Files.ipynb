## Opening a File

- Mode: string specifying how the file will be opened

- If open function receives a filename that does not contain a path, assumes that file is in same directory as program

- If program is running and file is created, it is created in the same directory as the program
    - Can specify alternative path and file name in the open function argument
        - Prefix the path string literal with the letter r
        test_file=open(r'C:\Users\Osama\temp\test.txt', 'w')


````python
# This program reads and displays the contents
# of the philosophers.txt file.
def main():
 
    # filename = 'philosophers.txt'        # English file
    infile = open("philosophers.txt", 'r')
    # infile = open("arabic-philosophers.txt", 'r')

    # infile = open("arabic-philosophers.txt", 'r')
    # Read the file's contents.
    file_contents = infile.read()

    # Close the file.
    infile.close()

    # Print the data that was read into memory.
    print(file_contents)

# Call the main function.
main()
````

## Reading Arabic (Unicode) Files
- Arabic text is stored using the **UTF-8** encoding
- Pass `encoding='utf-8'` to the open function so Python decodes the characters correctly
    - Format: 	file_object = open(filename, mode, encoding='utf-8')
- Without it, Windows may use a default encoding (cp1252) and you get garbled text or a `UnicodeDecodeError`


````python
# This program reads and displays the contents of a file.
# Select the file to open by changing the filename below.
def main():
    # filename = 'arabic-philosophers.txt'   # Arabic file
    filename = 'philosophers.txt'        # English file

    # Open the file using UTF-8 encoding (needed for Arabic text).
    infile = open(filename, 'r', encoding='utf-8')

    # Read the file's contents.
    file_contents = infile.read()

    # Close the file.
    infile.close()

    # Print the data that was read into memory.
    print(file_contents)

# Call the main function.
main()
````

## Reading Numeric Data from a File
- Number are read from a text file as strings
    - Must be converted to numeric type in order to perform mathematical operations
    - Use int and float functions to convert string to numeric value


````python
# This program demonstrates how numbers that are
# read from a file must be converted from strings
# before they are used in a math operation.


def main():
    # Open a file for reading.
    myfile = open('numbers.txt', 'r')

    # Read three numbers from the file.
    num1 = int(myfile.readline()) ## casting to int convert to int
    num2 = int(myfile.readline())
    num3 = int(myfile.readline())

    # Close the file.
    myfile.close()

    # Add the three numbers.
    total = num1 + num2 + num3
    avg = total / 3

    # Display the numbers and their total.
    print(f'The numbers are: {num1}, {num2}, {num3}')
    print(f'Their total is: {total}')
    print(f'Their average is: {avg}')

# Call the main function.
main()
````

## Using Loops to Process Files
- Files typically used to hold large amounts of data
    - Loop typically involved in reading from and writing to a file
- Often the number of items stored in file is unknown
    - The readline method uses an empty string as a sentinel when end of file is reached
    - Can write a while loop with the condition 	`while line != ''`


````python
# This program reads all of the values in
# the sales.txt file.

def main():
    # Open the sales.txt file for reading.
    sales_file = open('sales.txt', 'r')

    # Read the first line from the file, but
    # don't convert to a number yet. We still
    # need to test for an empty string.
    line = sales_file.readline()
    total_amount = 0.0
    # As long as an empty string is not returned
    # from readline, continue processing.
    while line != '':
        # Convert line to a float.
        amount = float(line)

        # Format and display the amount.
        print(f'{amount:.2f}')
        total_amount += amount

        # Read the next line.
        line = sales_file.readline()
        
    # Close the file.
    sales_file.close()
    print(f'The total sales amount is: {total_amount:.2f}')

# Call the main function.
main()
````

## Using Python’s for Loop to Read Lines
- Python allows the programmer to write a for loop that automatically reads lines in a file and stops when end of file is reached
    - Format: 	`for line in file_object:
                        statements`
    - The loop iterates once over each line in the file


````python
# This program uses the for loop to read
# all of the values in the sales.txt file.

def main():
    total = 0.0
    # Open the sales.txt file for reading.
    sales_file = open('sales.txt', 'r')

    # Read all the lines from the file.
    for line in sales_file:
        # Convert line to a float.
        amount = float(line)
        total+= amount
        # Format and display the amount.
        print(f'{amount:.2f}')
        
    # Close the file.
    sales_file.close()
    print(f'The total is: {total:.2f}')

# Call the main function.
main()
````

## Processing Records
- Record: set of data that describes one item
- Field: single piece of data within a record
- Write record to sequential access file by writing the fields one after the other
- Read record from sequential access file by reading each field until record complete
- When working with records, it is also important to be able to:
    - Add records
    - Display records
    - Search for a specific record
    - Modify records


````python
# This program gets employee data from the user and
# saves it as records in the employee.txt file.

def main():
    # Get the number of employee records to create.
    num_emps = int(input('How many employee records do you want to create? '))

    # Open a file for writing.
    emp_file = open('employees.txt', 'w')

    # Get each employee's data and write it to the file.
    for count in range(1, num_emps + 1):
        # Get the data for an employee.
        print(f'Enter data for employee #{count}')
        name = input('Name: ')
        id_num = input('ID number: ')
        dept = input('Department: ')

        # Write the data as a record to the file.
        emp_file.write(f'{name}\n')
        emp_file.write(f'{id_num}\n')
        emp_file.write(f'{dept}\n')
        
        # emp_file.write(f'{name}\n{id_num}\n{dept}\n')

        # Display a blank line.
        # print()

    # Close the file.
    emp_file.close()
    print('Employee records written to employees.txt.')

# Call the main function.
main()
````

````python
# This program displays the records that are
# in the employees.txt file.

def main():
    # Open the employees.txt file.
    emp_file = open('employees.txt', 'r')

    # Read the first line from the file, which is
    # the name field of the first record.
    name = emp_file.readline()

    # If a field was read, continue processing.
    while name:       # while name:
        # Read the ID number field.
        id_num = emp_file.readline()

        # Read the department field.
        dept = emp_file.readline()

        # Strip the newlines from the fields.
        name = name.rstrip('\n')
        id_num = id_num.rstrip('\n')
        dept = dept.rstrip('\n')

        # Display the record.
        print(f'Name: {name}')
        print(f'ID: {id_num}')
        print(f'Dept: {dept}')
        # print()

        # Read the name field of the next record.
        name = emp_file.readline()

    # Close the file.
    emp_file.close()

# Call the main function.
main()
````

## Exception Handling

- An **exception handler** is code that responds when an exception is raised and prevents the program from crashing.
    - In Python, handlers are written using a `try`/`except` statement.

- General format:

    ```python
    try:
            statements
    except ExceptionName:
            statements
    ```

- The **try suite** contains statements that may raise an exception.
- The **handler** contains the statements in the `except` block.

- If a statement in the `try` suite raises an exception:
    - If the exception matches the `except` clause, its handler runs, and the program continues after the `try`/`except` statement.
    - If it does not match, the program halts and displays a traceback.

- If no exception is raised, the handler is skipped.


## Displaying an Exception’s Default Error Message

- An **exception object** is created when an exception is raised.
    - It usually contains a default error message.
- Assign the exception object to a variable in an `except` clause:
    - `except ValueError as err:`
- Pass the exception object to `print()` to display its default error message.


````python
# This program calculates gross pay.

def main():
    try:
        # Get the number of hours worked.
        hours = int(input('How many hours did you work? '))

        # Get the hourly pay rate.
        pay_rate = float(input('Enter your hourly pay rate: '))

        # Calculate the gross pay.
        gross_pay = hours * pay_rate

        # Display the gross pay.
        print('Gross pay: $', format(gross_pay, ',.2f'), sep='')
    except ValueError as err:
        print(err)

# Call the main function.
main()
````

## The `else` Clause

- A `try`/`except` statement may include an optional `else` clause.
    - It appears after all `except` clauses and is aligned with `try` and `except`.
    - Its syntax is similar to the `else` clause in a decision structure.
- The `else` suite runs only if no exception is raised in the `try` suite.
    - If an exception is raised, the `else` suite is skipped.


````python
# This program displays the total of the amounts in the sales.txt file.

def main():
    total = 0.0	# Initialize an accumulator.
        
    try:
        infile = open('salesx.txt', 'r')     # Open the sales_data.txt file.

        for line in infile:	# Read the values from the file and accumulate them.
            amount = float(line)
            total += amount

        # infile.close()	  # Close the file.
        
    except Exception as err:
        print(err)
    else:
        print(f'{total:,.2f}')	 # Print the total.
    finally:
            print("code clean up....")
            # file.close()
        
main()	# Call the main function.
````

### Exercise: Handling File Operations and Exceptions

**Objective:** Write a Python program that reads data from a file, processes the data, and handles any potential exceptions that may occur during file operations.

**Instructions:**

1. Create a text file named `data.txt` and add the following lines to it:
    ```
    25
    30
    45
    50
    60
    ```

2. Write a Python program that performs the following tasks:
    - Open the `data.txt` file for reading.
    - Read each line from the file and convert it to an integer.
    - Calculate the sum and average of the numbers.
    - Handle any potential exceptions that may occur during file operations, such as `FileNotFoundError`, `ValueError`, and `ZeroDivisionError`.
    - Display appropriate error messages for each exception.
    - Ensure the file is properly closed after reading.

**Example Output:**
```
Sum of numbers: 210
Average of numbers: 42.0
```



**Example Error Handling:**


````python
Error: The file 'data.txt' was not found.
Error: Could not convert data to an integer.
Error: Division by zero occurred while calculating the average.
````



**Hint:** Use `try`-`except` blocks to handle exceptions and a `with` statement to ensure the file is properly closed.

**Solution:**



````python
def main():
    try:
        # Open the data.txt file for reading.
        # file = open('data.txt', 'r') #you must closte it
        with open('data.txt', 'r') as file:
            numbers = [] #list to store numbers
            
            # Read each line from the file.
            for line in file:
                try:
                    # Convert the line to an integer and add to the list.
                    number = int(line.strip())
                    numbers.append(number)
                except ValueError:
                    print(f"Error: Could not convert '{line.strip()}' to an integer.")
            
            # Calculate the sum and average of the numbers.
            total = sum(numbers)
            average = total / len(numbers)
            
            # Display the results.
            print(f"Sum of numbers: {total}")
            print(f"Average of numbers: {average:.2f}")
    
    except FileNotFoundError:
        print("Error: The file 'data.txt' was not found.")
    except ZeroDivisionError:
        print("Error: Division by zero occurred while calculating the average.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

# Call the main function.
main()
````
